```python
class Manusia:
    def __init__(self, tangan, kaki):
        self.tangan = tangan
        self.kaki = kaki
    
    def welcome(self):
        print(f"{self.tangan} {self.kaki}")

class Hewan(Manusia):
    def __init__(self, tangan, kaki, ekor):
        super().__init__(tangan, kaki)
        self.ekor = ekor
    
    def welcome(self):
        print(f"{self.tangan} {self.kaki} {self.ekor}")

manusia = Manusia("tangan manusia", "kaki manusia")
hewan = Hewan("tangan hewan", "kaki hewan", "ekor hewan")

manusia.welcome()
hewan.welcome()
```
